```python
import json
import pandas as pd

def convert_c3_to_excel(json_file_path, output_excel_path):
    # 讀取原始 JSON 檔案
    with open(json_file_path, 'r', encoding='utf-8') as f:
        data = json.load(f)

    rows = []
    current_id = 1

    # 遍歷 C3 資料集結構
    for entry in data:
        # entry[0] 通常是文章內容 (Article)，為一個字串列表
        article_text = "\n".join(entry[0])
        
        # entry[1] 是問題列表，可能包含多個問題
        questions_list = entry[1]
        
        for q_item in questions_list:
            question = q_item.get('question', '')
            choices = q_item.get('choice', [])
            correct_answer_text = q_item.get('answer', '')
            
            # 處理選項，不夠 4 個補空值
            opt1 = choices[0] if len(choices) > 0 else ""
            opt2 = choices[1] if len(choices) > 1 else ""
            opt3 = choices[2] if len(choices) > 2 else ""
            opt4 = choices[3] if len(choices) > 3 else ""
            
            # 將答案文字轉換為數字索引 (1, 2, 3, 4)
            try:
                answer_index = choices.index(correct_answer_text) + 1
            except ValueError:
                answer_index = None  # 若找不到匹配答案則設為空
            
            # 建立單一列數據
            rows.append({
                "ID": current_id,
                "Article": article_text,
                "Question": question,
                "Option1": opt1,
                "Option2": opt2,
                "Option3": opt3,
                "Option4": opt4,
                "Answer": answer_index
            })
            current_id += 1

    # 轉換為 DataFrame 並儲存
    df = pd.DataFrame(rows)
    df.to_excel(output_excel_path, index=False)
    print(f"轉換完成！檔案已儲存至: {output_excel_path}")

# d	Dialogue (對話)	以對話為主的閱讀理解題。	DREAM
# m	Mixed-genre (混合)	包含短文、信件、新聞等多種正式文體的混合類型。	RACE
convert_c3_to_excel('c3-m-train.json', 'c3-m-train.xlsx')
convert_c3_to_excel('c3-m-dev.json',   'c3-m-dev.xlsx')
convert_c3_to_excel('c3-m-test.json',  'c3_m_test.xlsx')
convert_c3_to_excel('c3-d-train.json', 'c3-d-train.xlsx')
convert_c3_to_excel('c3-d-dev.json',   'c3-d-dev.xlsx')
convert_c3_to_excel('c3-d-test.json',  'c3_d_test.xlsx')
```
