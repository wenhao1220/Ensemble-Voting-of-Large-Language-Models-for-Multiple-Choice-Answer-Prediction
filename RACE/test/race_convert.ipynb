```python
import os
import json
import pandas as pd

def process_race_data(folder_path, output_file):
    all_data = []
    serial_number = 1
    
    # 建立答案映射字典
    answer_map = {'A': 1, 'B': 2, 'C': 3, 'D': 4}
    
    # 遍歷資料夾中的所有檔案
    for filename in os.listdir(folder_path):
        if filename.endswith(".txt"):
            file_path = os.path.join(folder_path, filename)
            
            with open(file_path, 'r', encoding='utf-8') as f:
                try:
                    data = json.load(f)
                    
                    # RACE 格式中，一個檔案可能包含多個問題
                    # 遍歷該檔案內的所有問題
                    for i in range(len(data['questions'])):
                        row = {
                            "ID": serial_number,
                            "Original_ID": data.get('id', filename),
                            "Article": data.get('article', ""),
                            "Question": data['questions'][i],
                            "Option1": data['options'][i][0] if len(data['options'][i]) > 0 else "",
                            "Option2": data['options'][i][1] if len(data['options'][i]) > 1 else "",
                            "Option3": data['options'][i][2] if len(data['options'][i]) > 2 else "",
                            "Option4": data['options'][i][3] if len(data['options'][i]) > 3 else "",
                            "Answer": answer_map.get(data['answers'][i], data['answers'][i])
                        }
                        all_data.append(row)
                        serial_number += 1
                except Exception as e:
                    print(f"Error processing {filename}: {e}")

    # 轉換成 DataFrame
    df = pd.DataFrame(all_data)
    
    # 匯出至 Excel
    df.to_excel(output_file, index=False)
    print(f"處理完成！共轉換 {len(df)} 筆題目，已儲存至 {output_file}")

# 使用範例 (請替換為你的實際路徑)
folder_path = './high'
output_file = 'RACE_High_Combined.xlsx'
process_race_data(folder_path, output_file)
```
