# 版本1： 移除標點符號，用ckiplab/bert-base-chinese-qa預測答案A，從4個選項(若答案不在題目中的不考慮)中找一個跟A最接近的答案當答案

```python
import pandas as pd
from transformers import pipeline
from text2vec import Similarity
import opencc
import tqdm
from datetime import datetime as dt
import re

def remove_punctuation(text):
    # 正則表達式：匹配所有中文與英文標點符號
    punctuation_pattern = r'[!"#$%&\'()*+,\-./:;<=>?@\[\\\]^_`{|}~，。、？！；：「」『』（）《》〈〉—…～﹏‧·]'
    return re.sub(punctuation_pattern, '', text)

model_name = "ckiplab/bert-base-chinese-qa"

# 建立問答 pipeline
AG = pipeline('question-answering', model=model_name, tokenizer=model_name)

# 定义匹配函数
sim_model = Similarity("shibing624/text2vec-base-multilingual")

def answer_multiple_choice_question(context, question, options, answer):
    result = AG(
        question = question,
        context = context
    )

    count = 0;
    
    for i in range(0,len(options)): 
        if str(options[i]).lower() in str(context).lower():
            count += 1
    
    score = -2

    if(count > 0):
        for i in range(0,len(options)): 
            if str(options[i]).lower() in str(context).lower():
                new_score = sim_model.get_score(result['answer'], options[i])
                if new_score > score:
                    predicted_answer = i+1
                    score = new_score
    else:
        for i in range(0,len(options)): 
            new_score = sim_model.get_score(result['answer'], options[i])
            if new_score > score:
                predicted_answer = i+1
                score = new_score

    is_correct = 1 if predicted_answer == answer else 0
    return is_correct, predicted_answer, result['answer']

# Load your Excel file with pandas
df = pd.read_excel('../data/all.xlsx')

# 新增兩個欄位以存放預測與模型抽取答案
df['predicted_answer'] = None
df['result_answer'] = None

ACC = 0

start_time = dt.now()

for i in tqdm.tqdm(range(len(df))):  # Use the full range len(df) in actual use
    try:
        now = df.iloc[i]
        
        context = str(now['Article'])
        question = str(now['Question'])
        options = [
            str(now["Option1"]),
            str(now["Option2"]),
            str(now["Option3"]),
            str(now["Option4"])
        ]
        answer = int(now["Answer"])
        
        is_correct, predicted_answer, extracted_answer = answer_multiple_choice_question(
            context, question, options, answer
        )

        # 計數
        ACC += is_correct

        # 寫回當前列的新欄位
        df.at[i, 'predicted_answer'] = predicted_answer
        df.at[i, 'result_answer'] = extracted_answer

        if (i+1) % 500 == 0:
            tmp_time = dt.now()
            exe_time = tmp_time - start_time
            print(f"程式碼執行時間: {exe_time}")
            print(f"[{str(dt.now()):.19}] 跑完{i+1}筆資料")
            print(f"目前準確率{ACC}題正確，{i+1-ACC}題錯誤 準確率{ACC/(i+1)}\n")
    except Exception as e:
        tmp_time = dt.now()
        exe_time = tmp_time - start_time
        print(f"程式碼執行時間: {exe_time}")
        print(f"[{str(dt.now()):.19}] 跑完{i}筆資料")
        print(f"共{ACC}題正確；{i - ACC}題錯誤")
        print(f"Accuracy: {ACC / i}\n")
        print(f"Error: {e}")
        time.sleep(5)

end_time = dt.now()
exe_time = end_time - start_time
print(f"程式碼執行時間: {exe_time}")
print(f"共{ACC}題正確；{len(df) - ACC}題錯誤")
print(f"Accuracy: {ACC / len(df)}")

# 輸出含新欄位的結果檔
df.to_excel('../data/all_with_predictions.xlsx', index=False)
print("已輸出：../data/all_with_predictions.xlsx")
```

```python
import re
import torch
import tqdm
import pandas as pd
from sklearn.metrics import classification_report, accuracy_score
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from string import Template

# ---------------------------
# 0) 裝置
# ---------------------------
device = "cuda:0" if torch.cuda.is_available() else "cpu"
print("device:", device)
torch.manual_seed(42)

# ---------------------------
# 1) 載入 mT5（可處理中文）
# ---------------------------
MODEL_NAME = "google/mt5-base"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=False)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME).to(device)
model.eval()

# ---------------------------
# 2) 中文提示模板
# ---------------------------
preamble = (
    "請根據下列文章內容回答單選題。"
    "請只輸出一個選項字母：A、B、C 或 D（不得輸出其他任何文字）。"
)
template = Template(
    "$preamble\n\n"
    "【文章】\n$Article\n\n"
    "【題目】\n$Question\n\n"
    "【選項】\nA) $a\nB) $b\nC) $c\nD) $d\n\n"
    "請只輸出 A、B、C 或 D。"
)

def format_input(df: pd.DataFrame, idx: int) -> str:
    return template.substitute(
        preamble=preamble,
        Article=df.loc[idx, "Article"],
        Question=df.loc[idx, "Question"],
        a=df.loc[idx, "Option1"],
        b=df.loc[idx, "Option2"],
        c=df.loc[idx, "Option3"],
        d=df.loc[idx, "Option4"],
    )

# ---------------------------
# 3) 單題推論
# ---------------------------
LETTER_MAP = {"A": 1, "B": 2, "C": 3, "D": 4}
LETTER_REGEX = re.compile(r"[ABCD]", re.IGNORECASE)

def predict_one(prompt: str, max_new_tokens: int = 2, max_source_len: int = 1024) -> str:
    with torch.inference_mode():
        enc = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=max_source_len)
        enc = {k: v.to(device) for k, v in enc.items()}
        out = model.generate(
            **enc,
            do_sample=False,     # 使用貪婪解碼
            num_beams=1,
            max_new_tokens=max_new_tokens,
        )
        text = tokenizer.decode(out[0], skip_special_tokens=True).strip()
    m = LETTER_REGEX.search(text.upper())
    return m.group(0) if m else "A"  # 保底

# ---------------------------
# 4) 直接在 df 上推論 + 評估
#    需要欄位：Article, Question, Option1..4, correct_answer(1/2/3/4)
# ---------------------------
pred_letters = []
for idx in tqdm.tqdm(df.index, desc="Predicting on df"):
    prompt = format_input(df, idx)
    pred_letters.append(predict_one(prompt))

df = df.copy()
df["Predicted_Results"] = pred_letters                     # 字母
df["pred_num"] = df["Predicted_Results"].str.upper().map(LETTER_MAP).fillna(1).astype(int)  # 1~4

y_true = df["Answer"].astype(int)
y_pred = df["pred_num"].astype(int)

print("\n[ALL df] Classification report")
print(classification_report(y_true, y_pred, digits=4, zero_division=0))
print(f"[ALL df] Accuracy: {accuracy_score(y_true, y_pred):.4f}")

# 如需輸出結果檔：
# df.to_csv("df_with_predictions.csv", index=False, encoding="utf-8-sig")
```
