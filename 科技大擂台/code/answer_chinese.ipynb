```python
import unicodedata

def longest_common_substrings(s1: str, s2: str):
    """
    回傳 (最長長度, 最長子字串列表[去重後排序])
    """
    # 正規化，避免全形/組合字等 Unicode 差異
    #s1 = unicodedata.normalize('NFC', s1)
    #s2 = unicodedata.normalize('NFC', s2)

    n, m = len(s1), len(s2)
    if n == 0 or m == 0:
        return 0, []

    # dp[j] 代表以 s1[:i] 與 s2[:j] 結尾的最長共同"連續"長度
    prev = [0] * (m + 1)
    best_len = 0
    best_set = set()

    for i in range(1, n + 1):
        cur = [0] * (m + 1)
        for j in range(1, m + 1):
            if s1[i - 1] == s2[j - 1]:
                cur[j] = prev[j - 1] + 1
                if cur[j] > best_len:
                    best_len = cur[j]
                    best_set = {s1[i - best_len:i]}
                elif cur[j] == best_len and best_len > 0:
                    best_set.add(s1[i - best_len:i])
        prev = cur

    return best_len, sorted(best_set)

# 範例
a = "我在臺灣大學讀資工系"
b = "資工系的課很多，我朋友也在臺灣大學讀資工系"
a = "若滿分為100，請問哪個是滿分"
b = "100"
length, substrings = longest_common_substrings(a, b)
print("最長長度：", length)        # 例如輸出 3
print("最長片段：", substrings)    # 例如輸出 ['資工系']
```

```python
import re
from cn2an import cn2an

# ===== 1) 你的核心轉換函式 =====
def parse_cn_or_percent(s: str):
    s = s.strip()
    m = re.match(r"百分之\s*(.+)", s)
    if m:  # 「百分之X」
        inner = m.group(1)
        num = _to_number(inner)
        value = float(num) / 100.0
    elif "%" in s:  # 「X%」
        s2 = s.replace("%", "").strip()
        num = _to_number(s2)
        value = float(num) / 100.0
    else:  # 純數字（中文/阿拉伯）
        value = float(_to_number(s))
    # 整數不加 .0
    return int(value) if value.is_integer() else value

def _to_number(token: str):
    # 嘗試阿拉伯 → 否則中文（smart 自動判斷繁簡/大小寫數字）
    try:
        return float(token)
    except ValueError:
        # 支援「負/负」與「點/点」等中文小數
        # cn2an 會處理「兩、壹、貳…」等多種數字寫法
        return float(cn2an(token, "smart"))

# ===== 2) 全文替換器 =====
# 你可以把想處理的型態拆成三個 pattern，依序替換，避免重覆處理
PAT_PERCENT_ZH = re.compile(
    r"百分之\s*([+\-]?[0-9\.零〇一二兩三四五六七八九十百千萬亿億兆點点壹貳參肆伍陸柒捌玖负負]+)"
)
PAT_PERCENT_SYM = re.compile(
    r"([+\-]?[0-9\.零〇一二兩三四五六七八九十百千萬亿億兆點点壹貳參肆伍陸柒捌玖负負]+)\s*%"
)
# 盡量避免吃到小數點左右的阿拉伯數字；只針對純中文數字片段
PAT_CN_NUM = re.compile(
    r"(?<![\d\.])([负負]?[零〇一二兩三四五六七八九十百千萬亿億兆壹貳參肆伍陸柒捌玖]+(?:[點点][零〇一二兩三四五六七八九壹貳參肆伍陸柒捌玖]+)?)"
    r"(?![\d\.%])"
)

def _fmt(value: float) -> str:
    # 整數 → int；浮點 → 去除多餘尾零與科學記號
    if isinstance(value, float) and value.is_integer():
        return str(int(value))
    return f"{value:.15g}"

def replace_cn_numbers(text: str) -> str:
    # 1) 「百分之X」 → 轉成小數
    def repl_percent_zh(m):
        whole = "百分之" + m.group(1)
        v = parse_cn_or_percent(whole)
        return _fmt(v)

    # 2) 「X%」 → 轉成小數
    def repl_percent_sym(m):
        whole = m.group(0)
        v = parse_cn_or_percent(whole)
        return _fmt(v)

    # 3) 純中文數字 → 阿拉伯數字
    def repl_cn_num(m):
        token = m.group(1)
        v = parse_cn_or_percent(token)
        return _fmt(v)

    # 依序替換，避免互相影響
    out = PAT_PERCENT_ZH.sub(repl_percent_zh, text)
    out = PAT_PERCENT_SYM.sub(repl_percent_sym, out)
    out = PAT_CN_NUM.sub(repl_cn_num, out)

    return out

# ===== 3) 範例 =====
if __name__ == "__main__":
    s = (
        "今年營收成長百分之五、毛利率約35%，"
        "去年的成長率是3%，本季目標百分之一百。"
        "樣品數量為一百二十三，誤差三點五。"
        "負十二與負三點五也要能解析。"
    )
    print(replace_cn_numbers(s))
```

```python
# -*- coding: utf-8 -*-
"""
Chinese number & percentage replacer
------------------------------------
pip install cn2an

提供兩個主要API：
1) parse_cn_or_percent(s): 將單一片段（中文數字 / X% / 百分之X / 千分之X / 萬分之X）轉成數值（int或float）。
2) replace_cn_numbers(text, **options): 對整段文字做就地替換。

作者：你
"""

import re
from typing import Iterable, Pattern, Tuple
from cn2an import cn2an

# ===== 0) 規則與正規化 =====

# 單獨位階字：自動補「一」（例如「百」→「一百」，避免 cn2an 拒絕）
UNIT_ONLY = {"十", "拾", "百", "佰", "千", "仟", "萬", "万", "億", "亿", "兆"}

# 中文數字字形正規化（可按需求增減）
CN_NORMALIZE = {
    "兩": "二", "俩": "二",
    "点": "點",
    "负": "負",
    "〇": "零", "○": "零", "Ｏ": "零",
    "拾": "十", "佰": "百", "仟": "千",
    "万": "萬", "亿": "億",
}

# 量級詞（分母）：百分/千分/萬分
DENOM_WORDS = {
    "百分之": 100.0,
    "千分之": 1000.0,
    "萬分之": 10000.0,
}

# ===== 1) 解析核心 =====

def _normalize_cn_token(token: str) -> str:
    t = token.strip()
    for k, v in CN_NORMALIZE.items():
        t = t.replace(k, v)
    if t in UNIT_ONLY:
        t = "一" + t  # 對單獨位階字補「一」
    return t

def _to_number(token: str) -> float:
    """
    嘗試將 token 轉成浮點數：
    1) 先嘗試阿拉伯數字
    2) 失敗則正規化中文，再用 cn2an 解析
    解析失敗會 raise 讓外部決定是否保留原字串
    """
    token = token.strip()
    # 先試阿拉伯
    try:
        return float(token)
    except ValueError:
        pass

    # 中文
    t = _normalize_cn_token(token)
    # 特例：一十 → 10；也可直接交給 cn2an 解析（它也能處理）
    try:
        return float(cn2an(t, "smart"))
    except Exception:
        raise

def _fmt_number(value: float) -> str:
    """整數不帶 .0，浮點去除多餘尾零，不用科學記號。"""
    if isinstance(value, float) and value.is_integer():
        return str(int(value))
    return f"{value:.15g}"

def parse_cn_or_percent(s: str):
    """
    將單一文字片段轉為數值：
    支援：
      - 純中文數字：一百二十三 → 123
      - 小數：三點五 → 3.5
      - 負數：負三點五 → -3.5
      - 百分比：百分之五十 / 50% → 0.5
      - 千分比：千分之五 → 0.005
      - 萬分比：萬分之五 → 0.0005
    回傳 int 或 float
    """
    s = s.strip()
    # 量級詞（百分/千分/萬分）
    for word, denom in DENOM_WORDS.items():
        m = re.match(rf"{word}\s*(.+)", s)
        if m:
            inner = m.group(1)
            num = _to_number(inner)
            value = float(num) / denom
            return int(value) if value.is_integer() else value

    # X%
    if "%" in s:
        # 允許 " 99.5 % "
        s2 = s.replace("%", "").strip()
        num = _to_number(s2)
        value = float(num) / 100.0
        return int(value) if value.is_integer() else value

    # 純數字（中文/阿拉伯）
    value = float(_to_number(s))
    return int(value) if value.is_integer() else value

# ===== 2) 全文替換器 =====

# 量級詞（百分/千分/萬分） → 小數
PAT_DENOM_ZH: Pattern = re.compile(
    r"(百分之|千分之|萬分之)\s*([+\-]?[0-9\.零〇一二兩三四五六七八九十百千萬亿億兆點点壹貳參肆伍陸柒捌玖负負]+)"
)

# X%
PAT_PERCENT_SYM: Pattern = re.compile(
    r"([+\-]?[0-9\.零〇一二兩三四五六七八九十百千萬亿億兆點点壹貳參肆伍陸柒捌玖负負]+)\s*%"
)

# 純中文數字（避免吃到鄰近阿拉伯或 %）；允許負號與小數
PAT_CN_NUM: Pattern = re.compile(
    r"(?<![\d\.])"                                  # 左側不是數字或小數點
    r"([负負]?[零〇一二兩三四五六七八九十百千萬亿億兆壹貳參肆伍陸柒捌玖]+"
    r"(?:[點点][零〇一二兩三四五六七八九壹貳參肆伍陸柒捌玖]+)?)"
    r"(?![\d\.%])"                                  # 右側不是數字/小數點/%
)

# 預設的語境排除（避免把日期、年分等錯轉）
DEFAULT_EXCLUSION_PATTERNS: Tuple[Pattern, ...] = (
    # 2025年、二〇二五年、二零二五年
    re.compile(r"(?:\d{2,4}|[零〇一二兩三四五六七八九]{2,4})\s*年"),
    # 一月/二月/十二月/週一/星期一
    re.compile(r"(?:[零〇一二兩三四五六七八九十]{1,3})(?:\s*[月號日週星期])"),
    # 時間（兩點/三點五分）
    re.compile(r"(?:[零〇一二兩三四五六七八九十]{1,3})\s*點"),
)

def _is_excluded(span: Tuple[int, int], text: str, exclusions: Iterable[Pattern]) -> bool:
    """判斷匹配區段是否位於排除語境（任一排除規則在該區段附近命中）。"""
    start, end = span
    window_l = max(0, start - 6)  # 粗略取上下文視窗
    window_r = min(len(text), end + 6)
    ctx = text[window_l:window_r]
    return any(p.search(ctx) for p in exclusions)

def replace_cn_numbers(
    text: str,
    *,
    apply_denoms: bool = True,
    apply_symbol_percent: bool = True,
    apply_pure_cn: bool = True,
    exclusions: Iterable[Pattern] = DEFAULT_EXCLUSION_PATTERNS,
) -> str:
    """
    對整段文字進行替換：
        - 百分/千分/萬分之X → 小數
        - X% → 小數
        - 純中文數字 → 阿拉伯數字
    參數：
        apply_denoms: 是否處理「百分/千分/萬分之」
        apply_symbol_percent: 是否處理「X%」
        apply_pure_cn: 是否處理純中文數字
        exclusions: 語境排除的正則清單
    """
    out = text

    # 1) 量級詞（百分/千分/萬分之）
    if apply_denoms:
        def repl_denoms(m):
            # 排除語境
            if _is_excluded(m.span(), out, exclusions):
                return m.group(0)
            whole = m.group(0)
            try:
                v = parse_cn_or_percent(whole)
                return _fmt_number(v)
            except Exception:
                return whole
        out = PAT_DENOM_ZH.sub(repl_denoms, out)

    # 2) X%
    if apply_symbol_percent:
        def repl_percent_sym(m):
            if _is_excluded(m.span(), out, exclusions):
                return m.group(0)
            whole = m.group(0)
            try:
                v = parse_cn_or_percent(whole)
                return _fmt_number(v)
            except Exception:
                return whole
        out = PAT_PERCENT_SYM.sub(repl_percent_sym, out)

    # 3) 純中文數字
    if apply_pure_cn:
        # 為避免替換過程中 span 移動導致誤判，使用 finditer + 手工拼接
        parts = []
        last = 0
        for m in PAT_CN_NUM.finditer(out):
            s, e = m.span()
            token = m.group(1)
            # 排除語境
            if _is_excluded((s, e), out, exclusions):
                continue
            parts.append(out[last:s])
            try:
                v = parse_cn_or_percent(token)
                parts.append(_fmt_number(v))
            except Exception:
                parts.append(token)  # 解析失敗保留原文
            last = e
        parts.append(out[last:])
        out = "".join(parts)

    return out

# ===== 3) 範例與測試 =====
if __name__ == "__main__":
    demo = (
        "今年營收成長百分之五、毛利率約35%，"
        "去年的成長率是3%，本季目標百分之一百。"
        "樣品數量為一百二十三，誤差三點五。"
        "負十二與負三點五也要能解析。"
        "另外，千分之五、萬分之二也要轉。"
        "日期如2025年、一月、星期一不應該被轉。"
        "單獨位階字：十、百、千、萬。"
    )
    print("原文：")
    print(demo)
    print("\n替換後：")
    print(replace_cn_numbers(demo))
```
